```python
import fitz
from pathlib import Path
import json
```

```python
pdf_path = Path("../../data/raw/test/1_1_2_3_one_two_three_syrup_120_ml.pdf")
pdf_path
```

```python
if pdf_path.exists():
    print("PDf Found")
else:
    print("PDF not found")    
```

```python
doc = fitz.open(pdf_path)

pages_data = []

for page_number, page in enumerate(doc):

    text = page.get_text()

    pages_data.append(
        {
            "page_number": page_number + 1,
            "characters": len(text),
            "text": text
        }
    )

pages_data
```

```python
for page in pages_data:
    print("=" * 50)
    print("PAGE:", page["page_number"])
    print("Characters:", page["characters"])
    print("=" * 50)

    print(page["text"][:2000])
```

```python
total_characters = sum(
    page["characters"]
    for page in pages_data
)

print("Total characters:", total_characters)


if total_characters > 100:
    print("Text layer detected ")
else:
    print("OCR may be required ")
```

```python
SECTION_FIXES = {
    "( ةلاعفلا ةداملاActive Ingredient):":
        "Active Ingredient(الماده الفعاله):",

    "( جالعلا تارشؤم / تامادختسالاIndications):":
        "Indications(مؤشرات العلاج):",

    "( ةماه تاريذحتWarnings):":
        "Warnings(التحذيرات):"
}

def fix_section_headers(text):
    for wrong, correct in SECTION_FIXES.item():
        text = text.replace(
            wrong,
            correct
        )
    return text    
```
