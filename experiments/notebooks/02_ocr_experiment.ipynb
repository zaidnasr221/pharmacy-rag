```python
from collections import Counter
from pathlib import Path
import json
import random
import unicodedata

import fitz
project_root = Path.cwd()

while (
    not (project_root / "pyproject.toml").exists()
    and project_root != project_root.parent
):
    project_root = project_root.parent

pdf_dir = project_root / "data" / "raw" / "drug_pdfs"

assert (project_root / "pyproject.toml").exists(), (
    "Project root not found"
)

assert pdf_dir.exists(), (
    f"PDF directory not found: {pdf_dir}"
)

all_pdf_paths = sorted(pdf_dir.glob("*.pdf"))

print("Project root:", project_root)
print("PDF directory:", pdf_dir)
print("PDF count:", len(all_pdf_paths))
```

```python
SECTION_FIXES = {
    "( ةلاعفلا ةداملاActive Ingredient):":
        "Active Ingredient(الماده الفعاله):",

    "( جالعلا تارشؤم / تامادختسالاIndications):":
        "Indications(مؤشرات العلاج):",

    "( ةماه تاريذحتWarnings):":
        "Warnings(التحذيرات):",
}


EXPECTED_HEADERS = {
    "active_ingredient":
        "Active Ingredient(الماده الفعاله):",

    "indications":
        "Indications(مؤشرات العلاج):",

    "warnings":
        "Warnings(التحذيرات):",
}
GENERAL_NOTE_MARKER = (
    "ملاحظة: هذه النسخة تحتوي على اسم الدواء"
)
```

```python
def extract_pdf_text(pdf_path):
    with fitz.open(pdf_path) as document:
        return "\n".join(
            page.get_text()
            for page in document
        )


def clean_extracted_text(text):
    # تحويل أشكال الحروف العربية إلى Unicode عادي.
    text = unicodedata.normalize("NFKC", text)

    # توحيد عناوين الأقسام.
    for wrong, correct in SECTION_FIXES.items():
        text = text.replace(wrong, correct)

    # حذف الملاحظة العامة الثابتة من المصدر.
    cleaned_lines = [
        line
        for line in text.splitlines()
        if GENERAL_NOTE_MARKER not in line
    ]

    return "\n".join(cleaned_lines)


def analyze_pdf(pdf_path):
    raw_text = extract_pdf_text(pdf_path)
    clean_text = clean_extracted_text(raw_text)

    unfixed_headers = [
        wrong
        for wrong in SECTION_FIXES
        if wrong in clean_text
    ]

    missing_sections = [
        field_name
        for field_name, header in EXPECTED_HEADERS.items()
        if header not in clean_text
    ]

    return {
        "file_name": pdf_path.name,

        "normalization_status": (
            "success"
            if not unfixed_headers
            else "failed"
        ),

        "data_status": (
            "complete"
            if not missing_sections
            else "incomplete"
        ),

        "missing_sections": missing_sections,
        "unfixed_headers": unfixed_headers,
        "characters": len(clean_text),
    }
```

```python
assert unicodedata.normalize("NFKC", "ﺗﻮﺟﺪ") == "توجد"

test_header = "( ةلاعفلا ةداملاActive Ingredient):"

assert clean_extracted_text(test_header) == (
    "Active Ingredient(الماده الفعاله):"
)

note_test_text = """
تحذير حقيقي
.ملاحظة: هذه النسخة تحتوي على اسم الدواء، المادة الفعالة، الاستخدامات، والتحذيرات فقط.
"""

clean_note_test = clean_extracted_text(note_test_text)

assert "تحذير حقيقي" in clean_note_test
assert GENERAL_NOTE_MARKER not in clean_note_test

print("Cleaning functions work correctly.")
```

```python
sample_size = 1000
random_seed = 42

pdf_paths = random.Random(random_seed).sample(
    all_pdf_paths,
    min(sample_size, len(all_pdf_paths)),
)

print("Selected files:", len(pdf_paths))
```

```python
results = []

for index, pdf_path in enumerate(pdf_paths, start=1):
    try:
        result = analyze_pdf(pdf_path)

    except Exception as error:
        result = {
            "file_name": pdf_path.name,
            "normalization_status": "error",
            "data_status": "error",
            "error": str(error),
        }

    results.append(result)

    if index % 100 == 0:
        print(f"Processed: {index}/{len(pdf_paths)}")

print("Test completed.")
```

```python
tested_count = len(results)

normalization_counts = Counter(
    result["normalization_status"]
    for result in results
)

data_counts = Counter(
    result["data_status"]
    for result in results
)

print("Tested:", tested_count)
print("Header normalization:", dict(normalization_counts))
print("Data completeness:", dict(data_counts))

for status, count in data_counts.items():
    percentage = count / tested_count
    print(f"{status}: {count} ({percentage:.1%})")
```

```python
problem_results = [
    result
    for result in results
    if (
        result["normalization_status"] != "success"
        or result["data_status"] != "complete"
    )
]

print("Problem files:", len(problem_results))

for result in problem_results[:20]:
    print(
        json.dumps(
            result,
            ensure_ascii=False,
            indent=2,
        )
    )
```

```python
missing_section_counts = Counter(
    section
    for result in results
    for section in result.get("missing_sections", [])
)

missing_patterns = Counter(
    tuple(result.get("missing_sections", []))
    for result in results
    if result.get("missing_sections")
)

normalization_failures = [
    result
    for result in results
    if result["normalization_status"] == "failed"
]

extraction_errors = [
    result
    for result in results
    if result["normalization_status"] == "error"
]

complete_files = [
    result
    for result in results
    if result["data_status"] == "complete"
]

incomplete_files = [
    result
    for result in results
    if result["data_status"] == "incomplete"
]

print("Total:", len(results))
print("Complete:", len(complete_files))
print("Incomplete:", len(incomplete_files))
print("Normalization failures:", len(normalization_failures))
print("Extraction errors:", len(extraction_errors))
print("Missing sections:", dict(missing_section_counts))
print("Missing patterns:", dict(missing_patterns))
```

```python
def parse_drug_text(clean_text, file_name):
    lines = [
        line.strip()
        for line in clean_text.splitlines()
        if line.strip()
    ]

    section_positions = {}

    for field_name, header in EXPECTED_HEADERS.items():
        if header in lines:
            section_positions[field_name] = lines.index(header)

    ordered_sections = sorted(
        (
            position,
            field_name,
        )
        for field_name, position in section_positions.items()
    )

    drug = {
        "drug_name": lines[0] if lines else "",
        "active_ingredient": "",
        "indications": "",
        "warnings": "",
        "source": {
            "file_name": file_name,
        },
    }

    for section_number, (start, field_name) in enumerate(
        ordered_sections
    ):
        if section_number + 1 < len(ordered_sections):
            end = ordered_sections[section_number + 1][0]
        else:
            end = len(lines)

        section_content = lines[start + 1:end]

        drug[field_name] = "\n".join(
            section_content
        ).strip()

    missing_sections = [
        field_name
        for field_name in EXPECTED_HEADERS
        if field_name not in section_positions
    ]

    extraction_metadata = {
        "status": (
            "complete"
            if not missing_sections
            else "incomplete"
        ),
        "missing_sections": missing_sections,
    }

    return drug, extraction_metadata
```

```python
test_pdf_path = (
    pdf_dir
    / "1000_sebofree_antidandruf_shampoo_110ml.pdf"
)

raw_text = extract_pdf_text(test_pdf_path)
clean_text = clean_extracted_text(raw_text)

drug, extraction_metadata = parse_drug_text(
    clean_text=clean_text,
    file_name=test_pdf_path.name,
)

print("DRUG JSON:")
print(
    json.dumps(
        drug,
        ensure_ascii=False,
        indent=2,
    )
)

print("\nEXTRACTION METADATA:")
print(
    json.dumps(
        extraction_metadata,
        ensure_ascii=False,
        indent=2,
    )
)
```

```python
expected_schema_keys = {
    "drug_name",
    "active_ingredient",
    "indications",
    "warnings",
    "source",
}

parsed_drugs = []
parser_results = []

for index, pdf_path in enumerate(pdf_paths, start=1):
    try:
        raw_text = extract_pdf_text(pdf_path)
        clean_text = clean_extracted_text(raw_text)

        drug, extraction_metadata = parse_drug_text(
            clean_text=clean_text,
            file_name=pdf_path.name,
        )

        assert set(drug) == expected_schema_keys

        parsed_drugs.append(drug)

        parser_results.append(
            {
                "file_name": pdf_path.name,
                "parser_status": "success",
                **extraction_metadata,
            }
        )

    except Exception as error:
        parser_results.append(
            {
                "file_name": pdf_path.name,
                "parser_status": "error",
                "error": str(error),
            }
        )

    if index % 100 == 0:
        print(f"Parsed: {index}/{len(pdf_paths)}")

print("Parser test completed.")
```

```python
parser_status_counts = Counter(
    result["parser_status"]
    for result in parser_results
)

extraction_status_counts = Counter(
    result.get("status", "error")
    for result in parser_results
)

empty_field_counts = Counter(
    field_name
    for drug in parsed_drugs
    for field_name in (
        "drug_name",
        "active_ingredient",
        "indications",
        "warnings",
    )
    if not drug[field_name]
)

print("Tested:", len(parser_results))
print("Parser status:", dict(parser_status_counts))
print("Extraction status:", dict(extraction_status_counts))
print("Empty fields:", dict(empty_field_counts))
```

```python
header_leaks = []

for drug in parsed_drugs:
    for field_name in (
        "active_ingredient",
        "indications",
        "warnings",
    ):
        leaked_headers = [
            header
            for header in EXPECTED_HEADERS.values()
            if header in drug[field_name]
        ]

        if leaked_headers:
            header_leaks.append(
                {
                    "file_name": drug["source"]["file_name"],
                    "field_name": field_name,
                    "leaked_headers": leaked_headers,
                }
            )

warnings_with_note = [
    drug
    for drug in parsed_drugs
    if "ملاحظة:" in drug["warnings"]
]

print("Header leaks:", len(header_leaks))
print(
    "Warnings containing the general note:",
    len(warnings_with_note),
)
```

```python
complete_drugs = [
    drug
    for drug in parsed_drugs
    if drug["active_ingredient"]
]

incomplete_drugs = [
    drug
    for drug in parsed_drugs
    if not drug["active_ingredient"]
]

review_random = random.Random(42)

review_samples = (
    review_random.sample(
        complete_drugs,
        min(3, len(complete_drugs)),
    )
    +
    review_random.sample(
        incomplete_drugs,
        min(3, len(incomplete_drugs)),
    )
)

for drug in review_samples:
    print("=" * 100)
    print(
        json.dumps(
            drug,
            ensure_ascii=False,
            indent=2,
        )
    )
```
