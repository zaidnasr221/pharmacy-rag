```python
import re

def remove_arabic_labels_keep_english(text: str) -> str:
    # يحول:
    # "المادة الفعالة (Active Ingredient):"
    # إلى:
    # "Active Ingredient:"
    
    pattern = r'[\u0600-\u06FF\s]+\( *([^)]+?) *\)\s*:'
    return re.sub(pattern, r'\1:', text)
```
